```python
# Demonstrating Basic data type
#.......>>>>>>>
# String-Patient Name
patient_name="REbort"

#int-Patient ID
patient_id=105

#Float-Body temperature(F)
body_temp=101.4

#Bool-Is patient admitted
Is_admitted=True

#print patient Details
print("-----Patient Record-----")
print("Patient Name :",patient_name)
print("Patient ID :",patient_id)
print("Body Temperature :",body_temp,"F")
print("Admitted :",Is_admitted)

#Example use 
if Is_admitted:
    print("Status:Patient is under hospital care.")
else:
    print("Status:Patient discharged - follow up required.")
```

```python
#Smart Home System

#Device Information 
device_name="Air Conditioner"  #str
device_id=2023001    #int
power_usage=3.75   # float
is_on=True   #bool
#Threshold
threshold_usage=5.0  #float

#Device Display Information
print("===SMART DEVICE STATUS===")
print("Device name :",device_name)
print("Device ID :",device_id)
print("Current Usage:",power_usage)
print("Device on :",is_on)

# check if energy usage exceeds threshold.
if power_usage > threshold_usage:
    print("Alert:High power consumption!")
else:
    print("power usage is within limit.")
```
